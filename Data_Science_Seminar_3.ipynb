```python
import pandas as pd

# 重新加载数据以确保变量存在
titanic = pd.read_excel('titanic.xlsx')

# 计算每个船舱等级（Pclass）的平均幸存率
pclass_survival = titanic.groupby('Pclass')['Survived'].mean().reset_index()
pclass_survival.columns = ['Pclass', 'Survival Rate']
print("不同船舱等级的幸存率：")
display(pclass_survival)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 绘制船舱等级与幸存率的关系柱状图
plt.figure(figsize=(8, 5))
sns.barplot(x='Pclass', y='Survival Rate', data=pclass_survival, palette='viridis')
plt.title('Survival Probability by Passenger Class (Pclass)')
plt.xlabel('Passenger Class (Pclass)')
plt.ylabel('Survival Probability')
plt.ylim(0, 1)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
# 计算每个登船港口（Embarked）的平均幸存率
embarked_survival = titanic.groupby('Embarked')['Survived'].mean().reset_index()
embarked_survival.columns = ['Embarked', 'Survival Rate']
print("不同登船港口的幸存率：")
display(embarked_survival)
```

```python
# 绘制登船港口与幸存率的关系柱状图
plt.figure(figsize=(8, 5))
sns.barplot(x='Embarked', y='Survival Rate', hue='Embarked', data=embarked_survival, palette='muted', legend=False)
plt.title('Survival Probability by Port of Embarkation (Embarked)')
plt.xlabel('Port of Embarkation')
plt.ylabel('Survival Probability')
plt.ylim(0, 1)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
